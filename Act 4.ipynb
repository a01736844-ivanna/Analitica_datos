```python
#Importamos librerías requeridas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Carga del archivo sin indice
df= pd.read_csv('listingsBoston.csv')
df
```

```python
#Verificamos información del DataFrame
df.info()
```

#### a) Los registros de 5 “host name” diferentes

```python
hosts = df['host_name'].dropna().unique()[:5]
filtro_a = df[df['host_name'].isin(hosts)]
filtro_a
```

#### b) Hosts que se unieron a Airbnb después del año 2020

```python
filtro_b = df[df['host_since'] >= "2020-01-01"]
filtro_b
```

#### c) Hosts que responden “within a day”

```python
filtro_c = df[df['host_response_time'] == "within a day"]
filtro_c
```

#### d) Filas múltiplos de 200 (200, 400, 600, …)

```python
filtro_d = df.iloc[200::200]
filtro_d
```

#### e) Tipos de cuarto: “Entire home” y “Private room”

```python
filtro_e = df[df['room_type'].isin(["Entire home/apt", "Private room"])]
filtro_e
```

#### f) Registros con a lo mucho 4 camas

```python
filtro_f = df[df['beds'] <= 4]
filtro_f
```

#### g) Registros que sean superhost

```python
filtro_g = df[df['host_is_superhost'] == "t"]
filtro_g
```

#### h) Columnas impares

```python
filtro_h = df.iloc[:, 1::2]
filtro_h
```

```python
filtro_a.to_csv("filtro_a.csv")
filtro_b.to_csv("filtro_b.csv")
filtro_c.to_csv("filtro_c.csv")
filtro_d.to_csv("filtro_d.csv")
filtro_e.to_csv("filtro_e.csv")
filtro_f.to_csv("filtro_f.csv")
filtro_g.to_csv("filtro_g.csv")
filtro_h.to_csv("filtro_h.csv")
```
