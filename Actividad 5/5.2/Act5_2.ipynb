```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .xlsx sin indice
data = pd.read_excel('Gastos y costos 20-23.xlsx')
data.head(5)
```

```python
#Eliminar primera fila o filas innecesarias
#Se eliminan pora poder tener los datos correctos
data1= pd.read_excel('Gastos y costos 20-23.xlsx', skiprows=5)
data1.head(5)
```

```python
valores_nulos=data1.isnull().mean()
valores_nulos
```

```python
#Realizamos una copia del dataframe 
data2=data1.copy()
```

```python
#Eliminar columnas innecesarias
#Estas dos columnas están en su mayoría vacías, por lo que es mejor eliminarlas que rellenarlas.
data2= data1.drop(['POLIZA', 'Columna1'], axis=1)
data2.head(2)
```

```python
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un STRING en  concreto 
#Ya que los valores nulos son el más de la mitad del df, se sustituye con no folio para no eliminar la columna completa. 
#No podemos inventar folios individuales, por lo que se escribe NO FOLIO
data2["FOLIO"] =data2["FOLIO"].fillna("No folio") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un STRING en  concreto 
#Se sustituye con el tipo de gasto más común
data2["GASTO"] =data2["GASTO"].fillna("COMPRA") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un valor numérico en  concreto 
#Ya que TC tiene la mayoría de los datos "1", 
data2["TC"]= data2["TC"].fillna(1) 
data2
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos con promedio o media
data2["IMPORTE"]=data2["IMPORTE"].fillna(round(data1["IMPORTE"].mean(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos con promedio o media
data2["IVA"]=data2["IVA"].fillna(round(data1["IVA"].mean(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un valor numérico en  concreto 
#El valor faltante es un combustible, los demás combustibles son del tipo I, así que se rellena con ese valor.
data2["TIPO"]= data2["TIPO"].fillna('I') 
data2.head(2)
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Convertir DataFrame a CSV
data2.to_csv("gastos2023_sin_nulos.csv")
```
