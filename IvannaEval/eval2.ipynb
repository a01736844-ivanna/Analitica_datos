```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .csv sin indice
df = pd.read_csv('cuentas_credicel.csv', encoding='latin1')
df.head()
```

```python
df = df.drop(columns=['Unnamed: 33', 'Unnamed: 34', 'Unnamed: 35', 'Unnamed: 36'])
df
```

```python
df.info()
```

```python
#Corroboramos valores nulos
valores_nulos=df.isnull().sum()
valores_nulos
```

```python
# Todas las columnas tipo fecha (datetime): Reemplazar nulos con el string: "01/01/2030 00:00"
df['fecha_ultimo_pago']= df['fecha_ultimo_pago'].fillna("01/01/2030 00:00")
df['fecha_pago_proximo']= df['fecha_pago_proximo'].fillna("01/01/2030 00:00")
```

```python
#Corroboramos valores nulos
valores_nulos=df.isnull().sum()
valores_nulos
```

```python
# Todas las columnas tipo string (object): Reemplazar nulos con el string: "Sin registro"
df['empresa']= df['empresa'].fillna("Sin registro")
df['status_cuenta']= df['status_cuenta'].fillna("Sin registro")
df['riesgo']= df['riesgo'].fillna("Sin registro") #Aunque tiene ceros, tambíen hay registris tipo string
df
```

```python
# Reemplazar valores string en columnas con ceros
df['razones_buro'] = pd.to_numeric(df['razones_buro'], errors='coerce').fillna(0)
df['razones_buro'] = df['razones_buro'].astype(int)
df['semana_actual'] = pd.to_numeric(df['semana_actual'], errors='coerce').fillna(0)
df['semana_actual'] = df['semana_actual'].astype(int)
df['codigo_postal'] = pd.to_numeric(df['codigo_postal'], errors='coerce').fillna(0)
df['codigo_postal'] = df['codigo_postal'].astype(int)
df.head()
```

```python
# Todas las columnas tipo numerico (int o float): Reemplazar nulos con el valor "Promedio"
df['puntos']= df['puntos'].fillna(df['puntos'].mean())
df['porc_enganche']= df['porc_enganche'].fillna(df['porc_enganche'].mean())
df['razones_buro']= df['razones_buro'].fillna(df['razones_buro'].mean())
df['semana_actual']= df['semana_actual'].fillna(df['semana_actual'].mean())
df['codigo_postal']= df['codigo_postal'].fillna(df['codigo_postal'].mean())
df
```

```python
#Corroboramos valores nulos
valores_nulos=df.isnull().sum()
valores_nulos
```

```python
#Convertir archivos filtrados a CSV
df.to_csv("Evaluación_4.csv")
```

```python
#Nos aseguramos que las columnas numericas si sean int 
df['precio'] = df['precio'].astype(int)
df['enganche'] = df['enganche'].astype(int)
df['descuento'] = df['descuento'].astype(int)
df['monto_financiado'] = df['monto_financiado'].astype(int)
df['monto_accesorios'] = df['monto_accesorios'].astype(int)
df['puntos'] = df['puntos'].astype(int)
df['porc_enganche'] = df['porc_enganche'].astype(int)
df['porc_tasa'] = df['porc_tasa'].astype(int)
df['score_buro'] = df['score_buro'].astype(int)

df.info()
```

```python
# Aplicar métodos (Desviación Estándar o Rango Intercuartílico) para reemplazar Outliers, graficar
# diagramas de caja por cada columna previamente a la sustitución de outliers.
cuantitativas = df.select_dtypes(include=['number'])
cualitativas = df.select_dtypes(exclude=['number'])
```

```python
#Realizamos diagrama de caja o bigote de cada columna del dataframe
fig = plt.figure(figsize =(15, 8))
cuantitativas.plot(kind='box', vert=False)
plt.title("Valores Atípicos del Dataframe")
plt.show() #dibujamos el diagrama
```

```python
#Método aplicando desviación estandar. Encuentro los valores extremos
#desv est detecta menos datos outliers que interquartilicos
y=cuantitativas
Limite_Superior= y.mean() + 3*y.std()
Limite_Inferior= y.mean() - 3*y.std()
print("Limite superior permitido", Limite_Superior)
print("Limite inferior permitido", Limite_Inferior)
```

```python
#Obtenemos datos y los outliers se convierten en nulos en el DataFrame
df1= cuantitativas[(y<=Limite_Superior)&(y>=Limite_Inferior)]
df1
```

```python
#Corroboramos valores nulos del dataframe4
valores_nulos=df1.isnull().sum()
valores_nulos
```

```python
#Reemplazamos valores atípicos (nulos) del dataframe con "mean"
#Realizamos una copia del dataframe
data_clean=df1.copy()
data_clean=data_clean.fillna(round(df1.mean(),1))
data_clean
```

```python
#Corroboramos valores nulos del dataframe LIMPIO
valores_nulos=data_clean.isnull().sum()
valores_nulos
```

```python
#Método aplicando Cuartiles. Encuentro cuartiles 0.25 y 0.75
y=cuantitativas

percentile25=y.quantile(0.25) #Q1
percentile75=y.quantile(0.75) #Q3
iqr= percentile75 - percentile25

Limite_Superior_iqr= percentile75 + 1.5*iqr
Limite_Inferior_iqr= percentile25 - 1.5*iqr
print("Limite superior permitido", Limite_Superior_iqr)
print("Limite inferior permitido", Limite_Inferior_iqr)
```

```python
#Obtenemos datos limpios del Dataframe
df1_iqr= cuantitativas[(y<=Limite_Superior_iqr)&(y>=Limite_Inferior_iqr)]
df1_iqr
```

```python
#Corroboramos valores nulos del dataframe LIMPIO
valores_nulos=df1_iqr.isnull().sum()
valores_nulos
```

```python
#Reemplazamos valores atípicos (nulos) del dataframe con "mean"
#Realizamos una copia del dataframe
df2_iqr=df1_iqr.copy()
df2_iqr=df2_iqr.fillna(round(df1_iqr.median(),1))
df2_iqr
```

```python
# Unimos el dataframe cuantitativo limpio con el dataframe cualitativo
Datos_limpios = pd.concat([cualitativas, df2_iqr], axis=1)
Datos_limpios
```

```python
#Corroboramos valores nulos del dataframe LIMPIO
valores_nulos=Datos_limpios.isnull().sum()
valores_nulos
```

```python
#Convertir archivos filtrados a CSV
Datos_limpios.to_csv("Evaluación_5.csv")
```
