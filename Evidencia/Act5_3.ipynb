```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .csv sin indice
data = pd.read_csv('BostonComplete.csv')
data["price"] = data["price"].replace('[\$,]', '', regex=True)
data["price"] = data["price"].astype(float)
data.head(2)
```

```python
#Corroboramos valores nulos
valores_nulos=data.isnull().mean()
valores_nulos
```

```python
#Realizamos una copia del dataframe 
data1=data.copy()
```

```python
#Eliminar columnas innecesarias
#Estas tres columnas están en su mayoría vacías, neighbourhood_overview no es tan relevante y está 49% vacía
data2= data1.drop(['neighborhood_overview', 'calendar_updated', 'neighbourhood_group_cleansed'], axis=1)
data2.head(2)
```

```python
#Sustituir valores nulos por un STRING en  concreto 
#Rellenamos con NO DESCRIPTION porque no hay forma de saber la descripción de una propiedad
data2["description"] =data2["description"].fillna("No description") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un STRING en  concreto 
#Rellenamos con Boston, MA. porque la mayoría de los hosts son de ahí y sería el "promedio" de los datos.
data2["host_location"] =data2["host_location"].fillna("Boston, MA")
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un STRING en  concreto 
#Rellenamos con No INFO porque no podemos saber los datos de los hosts.
data2["host_about"] =data2["host_about"].fillna("No information")
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia adelante forward fill" ("ffill")
#Filtro por columnas
data2["host_response_time"] =data2["host_response_time"].fillna(method="ffill") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia atrás backward fill" ("bfill")
# Como son %, es muy complicado quitar el signo y cambiar de string a int.
data2["host_response_rate"] =data2["host_response_rate"].fillna(method="bfill") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia adelante forward fill" ("ffill")
#Filtro por columnas
data2["host_acceptance_rate"] =data2["host_acceptance_rate"].fillna(method="ffill")  
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un STRING en  concreto 
#Se sustituye con el tipo de host más común
data2["host_is_superhost"] =data2["host_is_superhost"].fillna("f") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un STRING en  concreto 
#El vecindario del host no es tan importante, por lo que se rellena con NON SPECIFIED
data2["host_neighbourhood"] =data2["host_neighbourhood"].fillna("Non specified") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un STRING en  concreto 
#Se sustituye con el unico string que se encuentra en la base de datos
data2["neighbourhood"] =data2["neighbourhood"].fillna("Neighbourhood highlights") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos con promedio o media
data2["bathrooms"]=data2["bathrooms"].fillna("0")
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos con promedio o media
data2["bedrooms"]=data2["bedrooms"].fillna(round(data1["bedrooms"].mean(),1))
```

```python
#Sustituir valores nulos con promedio o media
data2["beds"]=data2["beds"].fillna(round(data1["beds"].mean(),1))
```

```python
#Sustituir valores nulos con promedio o media
#Al inicio quité el signo de $ de price, así que puedo usar mean para rellenar
data2["price"]=data2["price"].fillna(round(data1["price"].mean(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un STRING en  concreto, en su mayoría es TRUE
data2["has_availability"] =data2["has_availability"].fillna("t") 
```

```python
#Sustituir valores nulos con promedio o media
data2["estimated_revenue_l365d"]=data2["estimated_revenue_l365d"].fillna(round(data1["estimated_revenue_l365d"].mean(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia adelante forward fill" ("ffill")
data2["first_review"] =data2["first_review"].fillna(method="ffill") 
```

```python
#Sustituir valores nulos por valores no nulos hacia adelante forward fill" ("ffill")
data2["last_review"] =data2["last_review"].fillna(method="ffill") 
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos con promedio o media
#Las últimas variables son todas numéricas, así que decidí usar la moda ya que reviews muy altas o muy bajas podrían afectar el promedio
data2["review_scores_rating"]=data2["review_scores_rating"].fillna(0)
data2["review_scores_accuracy"]=data2["review_scores_accuracy"].fillna(0)
data2["review_scores_cleanliness"]=data2["review_scores_cleanliness"].fillna(0)
data2["review_scores_checkin"]=data2["review_scores_checkin"].fillna(0)
data2["review_scores_communication"]=data2["review_scores_communication"].fillna(0)
data2["review_scores_location"]=data2["review_scores_location"].fillna(0)
data2["review_scores_value"]=data2["review_scores_value"].fillna(0)
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un STRING en  concreto, 
# no se puede saber el tipo de licencia, se llena con NON specified
data2["license"] =data2["license"].fillna("Non specified") 
```

```python
#Sustituir valores nulos con promedio o media
# Ya que son valores numéricos, se usa el promedio
data2["reviews_per_month"]=data2["reviews_per_month"].fillna(round(data1["reviews_per_month"].mean(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```
